# 20241114 奖励法

目前实验结果较好的两种奖励

1. 比例法：
$$
\begin{align*}
    \Delta_\text{comfort} &= \frac{\text{comfort}_\text{curr} - \text{comfort}_\text{prev}}{\text{comfort}_\text{prev}} \\
    \Delta_\text{elec\_saving} &= -\frac{\text{elec}_\text{prev} - \text{elec}_\text{curr}}{\text{elec}_\text{curr}} \\
    \text{reward} &= \frac{\Delta_\text{comfort}}{\Delta_\text{elec\_saving}}
\end{align*}
$$

2. 向量法：
$$
\begin{align*}
    \Delta_{\text{comfort}} &= \text{norm\_comfort}_\text{curr} - \text{norm\_comfort}_\text{prev} \\
    \Delta_{\text{elec}} &= \text{norm\_elec}_\text{curr} - \text{norm\_elec}_\text{prev} \\
    \text{angle} &= \arctan2(\Delta_{\text{comfort}}, \Delta_{\text{elec}}) \\
    \text{distance} &= \sqrt{\Delta_{\text{elec}}^2 + \Delta_{\text{comfort}}^2} \\ 
    \text{reward\_angle} &= \cos\left(\text{angle} - \frac{3\pi}{4}\right) \\
    \text{reward} &= \text{reward\_angle} \cdot \text{distance}
\end{align*}
$$


```python
import pandas
import plotly
df_train_baseline = pandas.read_csv("tmp/records_baseline.csv")
df_train_baseline_t_22 = pandas.read_csv('tmp/records_baseline_t_22.csv')
df_train_baseline_t_22_5 = pandas.read_csv('tmp/records_baseline_t_22.5.csv')
df_train_baseline_t_23 = pandas.read_csv('tmp/records_baseline_t_23.csv')
df_train_baseline_t_23_5 = pandas.read_csv('tmp/records_baseline_t_23.5.csv')
df_train = pandas.read_csv("datasave/records_train_HVAC-1FWEST.csv")
df_train['🕰️'] = pandas.to_datetime(df_train['🕰️'])
df_train_baseline_t_22['🕰️'] = pandas.to_datetime(df_train_baseline_t_22['🕰️'], format='mixed')
df_train_baseline_t_22_5['🕰️'] = pandas.to_datetime(df_train_baseline_t_22_5['🕰️'], format='mixed')
df_train_baseline_t_23['🕰️'] = pandas.to_datetime(df_train_baseline_t_23['🕰️'], format='mixed')
df_train_baseline_t_23_5['🕰️'] = pandas.to_datetime(df_train_baseline_t_23_5['🕰️'], format='mixed')
```

```python
import pandas as pd
import glob

# Directories containing the CSV files
directories = [
    'datasave/20241120/records_single_agent_angle_with_distance',
    'datasave/20241120/records_train_slope_angle',
    'datasave/20241120/records_train_slope_angle_diff_saving',
    'datasave/20241120/records_train_slope_angle_with_distance',
]

# Dictionaries to store averages per directory
directory_avg_pmv = {}
directory_avg_elec = {}

for dir_path in directories:
    df_avg_pmv = {}
    df_avg_elec = {}
    pmv_total = 0
    elec_total = 0
    file_count = 0

    # Get all CSV files in the directory
    csv_files = glob.glob(f"{dir_path}/*.csv")
    
    for file_path in csv_files:
        # Extract the room_id from the file name
        # Adjust this based on your actual file naming convention
        file_name = file_path.split('/')[-1]
        room_id = file_name.split('-')[-1].replace('.csv', '')
        
        # Read the CSV file
        df = pd.read_csv(file_path)
        # Filter out rows where 'occupancy' is 0
        df = df[df['occupancy'] != 0]
        
        # Calculate mean absolute PMV and mean electricity usage
        mean_abs_pmv = df['pmv'].abs().mean()
        mean_elec = df['elec'].mean()
        
        # Store the results for each room
        df_avg_pmv[room_id] = mean_abs_pmv
        df_avg_elec[room_id] = mean_elec
        
        # Accumulate totals for directory averages
        pmv_total += mean_abs_pmv
        elec_total += mean_elec
        file_count += 1

    # Calculate averages for the current directory
    pmv_avg = pmv_total / file_count if file_count else 0
    elec_avg = elec_total / file_count if file_count else 0

    # Store the averages per directory
    directory_avg_pmv[dir_path] = pmv_avg
    directory_avg_elec[dir_path] = elec_avg

    print(f"Directory: {dir_path}")
    print(f"Average PMV: {pmv_avg}")
    print(f"Average Electricity Usage: {elec_avg}\n")
```

```python
import numpy as np
import plotly.graph_objs as go

# 假设 df_multi_agent_slope_angle 已经定义
df = df_train_baseline[['elec', 'pmv']].copy()

# 计算 'comfort' 并归一化 'elec'
df['comfort'] = (0.5 - df['pmv'].abs()) / 0.5
df['elec_norm'] = df['elec'] / df['elec'].max()

# 计算连续点之间的差异
df['delta_elec_norm'] = df['elec_norm'].diff()
df['delta_comfort'] = df['comfort'].diff()

# 删除由于 diff() 导致的第一个 NaN 值
df = df.dropna().reset_index(drop=True)

# 在直角坐标系中绘制相对位置
fig = go.Figure(
    data=[
        go.Scatter(
            x=df['delta_elec_norm'],
            y=df['delta_comfort'],
            mode='markers',
            marker=dict(
                size=8,
                color=np.sqrt(df['delta_elec_norm']**2 + df['delta_comfort']**2),
                colorscale='Viridis',
                showscale=True
            ),
            text=df.index  # 可选：添加标签或提示信息
        )
    ],
    layout=go.Layout(
        title="基于上一个时间点的相对位置（baseline）",
        xaxis=dict(title="电力消耗变化（delta_elec_norm）"),
        yaxis=dict(title="舒适度变化（delta_comfort）")
    )
)

fig.show()
```
