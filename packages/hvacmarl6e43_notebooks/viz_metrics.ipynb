```python
from hvacmarl6e43.buildings import OfficeBuilding
from hvacmarl6e43.experiments import rl_experiments

metrics = {
    name: experiment.get_results().metrics_train
    for name, experiment in rl_experiments.items()
}
```

```python
import holoviews as hv

hv.extension('matplotlib', 'bokeh')
hv.extension('bokeh')

import numpy as np
```

```python
list(metrics['multiagent'].columns)
```

```python
hv.Overlay([
    hv.Curve(
        metrics['monoagent']['info/learner/default_policy/learner_stats/kl'],
        label='monoagent',
    ),
    hv.Curve(
        metrics['multiagent'][[
            f'info/learner/{zone_key}/learner_stats/kl' 
            for zone_key in OfficeBuilding.zone_keys
        ]].mean(axis='columns'),
        label='multiagent',
    ),
]).opts(xlabel='Training Iteration', ylabel='KL Divergence')
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
monoagent_kl = metrics['monoagent']['info/learner/default_policy/learner_stats/kl']

multiagent_kl = metrics['multiagent'][
    [f'info/learner/{zone_key}/learner_stats/kl' for zone_key in OfficeBuilding.zone_keys]
].mean(axis='columns')

plt.figure(figsize=(10, 6))

plt.plot(monoagent_kl.index, monoagent_kl.values, label='monoagent')

plt.plot(multiagent_kl.index, multiagent_kl.values, label='multiagent')

plt.xlabel('Training Iteration')
plt.ylabel('KL Divergence')
plt.legend()
plt.tight_layout()
# plt.savefig('KL Divergence.svg', format='svg')
plt.show()
```

```python
hv.Overlay([
    hv.Curve(
        metrics['monoagent']['info/learner/default_policy/learner_stats/total_loss'],
        label='monoagent',
    ),
    hv.Curve(
        metrics['multiagent'][[
            f'info/learner/{zone_key}/learner_stats/total_loss' 
            for zone_key in OfficeBuilding.zone_keys
        ]].mean(axis='columns'),
        label='multiagent',
    ),
]).opts(xlabel='Training Iteration', ylabel='Total Loss (Avg.)')
```

```python
rewards = {
    'monoagent':
        metrics['monoagent']['env_runners/policy_reward_mean/default_policy']
        .dropna(),
    'multiagent': 
        metrics['multiagent'][[
            f'env_runners/hist_stats/policy_{zone_key}_reward' 
            for zone_key in OfficeBuilding.zone_keys
        ]]
        .dropna()
        .map(np.mean)
        .mean(axis='columns'),
}
```

```python
hv.Layout([
    hv.Overlay([
        hv.Curve(
            reward_hist,
            label=label,
        )
        for label, reward_hist in rewards.items()
    ]).opts(xlabel='Training Iteration', ylabel='Reward'),
    # hv.Overlay([
    #     hv.Curve(
    #         reward_hist.cumsum(),
    #         label=label,
    #     )
    #     for label, reward_hist in rewards.items()
    # ]).opts(xlabel='Training Iteration', ylabel='Reward (Cumulative)'),
]).opts(shared_axes=False)
```

```python
rewards
print(type(rewards['monoagent']))
```

```python
plt.figure(figsize=(10, 6))
colors = {
    "monoagent": "#1f77b4",  # Blue (from tab10)
    "multiagent": "#ff7f0e", # Orange (from tab10)
}
for label, reward_hist in rewards.items():
    x = reward_hist.index
    y = reward_hist.values
    plt.plot(x,y, label=label,color=colors[label])
    avg_reward = reward_hist.mean()
    plt.axhline(y=avg_reward, color=colors[label], linestyle='--', 
                label=f"{label} average ({avg_reward:.4f})")
plt.xlabel('Training Iteration')
plt.ylabel('Reward')
plt.legend()
plt.tight_layout()
plt.savefig('reward.svg', format='svg')
plt.show()
```
