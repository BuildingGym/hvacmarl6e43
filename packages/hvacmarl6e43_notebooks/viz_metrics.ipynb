```python
from hvacmarl6e43.buildings import OfficeBuilding
from hvacmarl6e43.experiments import rl_experiments
from pathlib import Path
from hvacmarl6e43_notebooks import notebook_storage_base

figurestorepath = Path(notebook_storage_base)
metrics = {
    name: experiment.get_results().metrics_train
    for name, experiment in rl_experiments.items()
}

```

```python
metrics
```

```python
name_edit = {
    'monoagent': 'SAPPO-PBT (Special Case)',
    'multiagent': 'MAPPO-PBT (Proposed)',
}

for name in list(metrics.keys()):
    if name =='monoagent':
        metrics[name_edit[name]] = metrics[name]
        del metrics[name]
    elif name == 'multiagent':
        metrics[name_edit[name]] = metrics[name]
        del metrics[name]
```

```python
metrics
```

```python
import holoviews as hv

hv.extension('matplotlib', 'bokeh')
hv.extension('bokeh')

import numpy as np
```

```python
hv.Overlay([
    hv.Curve(
        metrics['SAPPO-PBT (Special Case)']['info/learner/default_policy/learner_stats/kl'],
        label='SAPPO-PBT (Special Case)',
    ),
    hv.Curve(
        metrics['MAPPO-PBT (Proposed)'][[
            f'info/learner/{zone_key}/learner_stats/kl' 
            for zone_key in OfficeBuilding.zone_keys
        ]].mean(axis='columns'),
        label='MAPPO-PBT (Proposed)',
    ),
]).opts(xlabel='Training Iteration', ylabel='KL Divergence')
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
colors = {
    'Baseline (23.0°C)': (239/255, 65/255, 67/255),  
    'Baseline (24.0°C)': (191/255, 30/255, 46/255),   
    'SAPPO-PBT (Special Case)': (115/255, 186/255, 214/255),     
    'MAPPO-PBT (Proposed)': (3/255, 50/255, 80/255),
}

monoagent_kl = metrics['SAPPO-PBT (Special Case)']['info/learner/default_policy/learner_stats/kl']

multiagent_kl = metrics['MAPPO-PBT (Proposed)'][
    [f'info/learner/{zone_key}/learner_stats/kl' for zone_key in OfficeBuilding.zone_keys]
].mean(axis='columns')

plt.figure(figsize=(7.5, 4))

plt.plot(monoagent_kl.index, monoagent_kl.values, label='SAPPO-PBT (Special Case)', color=colors['SAPPO-PBT (Special Case)'])

plt.plot(multiagent_kl.index, multiagent_kl.values, label='MAPPO-PBT (Proposed)', color=colors['MAPPO-PBT (Proposed)'])

plt.xlabel('Training Iteration', fontsize=12)
plt.ylabel('KL Divergence', fontsize=12)
plt.legend()
plt.tight_layout()
plt.savefig(figurestorepath /'KL_Divergence.svg', format='svg')
plt.show()
```

```python
hv.Overlay([
    hv.Curve(
        metrics['SAPPO-PBT (Special Case)']['info/learner/default_policy/learner_stats/total_loss'],
        label='SAPPO-PBT (Special Case)',
    ),
    hv.Curve(
        metrics['MAPPO-PBT (Proposed)'][[
            f'info/learner/{zone_key}/learner_stats/total_loss' 
            for zone_key in OfficeBuilding.zone_keys
        ]].mean(axis='columns'),
        label='MAPPO-PBT (Proposed)',
    ),
]).opts(xlabel='Training Iteration', ylabel='Total Loss (Avg.)')
```

```python
rewards = {
    'SAPPO-PBT (Special Case)':
        metrics['SAPPO-PBT (Special Case)']['env_runners/policy_reward_mean/default_policy']
        .dropna(),
    'MAPPO-PBT (Proposed)': 
        metrics['MAPPO-PBT (Proposed)'][[
            f'env_runners/hist_stats/policy_{zone_key}_reward' 
            for zone_key in OfficeBuilding.zone_keys
        ]]
        .dropna()
        .map(np.mean)
        .mean(axis='columns'),
}
```

```python
hv.Layout([
    hv.Overlay([
        hv.Curve(
            reward_hist,
            label=label,
        )
        for label, reward_hist in rewards.items()
    ]).opts(xlabel='Training Iteration', ylabel='Reward'),
    # hv.Overlay([
    #     hv.Curve(
    #         reward_hist.cumsum(),
    #         label=label,
    #     )
    #     for label, reward_hist in rewards.items()
    # ]).opts(xlabel='Training Iteration', ylabel='Reward (Cumulative)'),
]).opts(shared_axes=False)
```

```python
plt.figure(figsize=(10, 6))
for label, reward_hist in rewards.items():
    x = reward_hist.index
    y = reward_hist.values
    plt.plot(x,y, label=label,color=colors[label])
    avg_reward = reward_hist.mean()
    plt.axhline(y=avg_reward, color=colors[label], linestyle='--', 
                label=f"{label} average ({avg_reward:.4f})")
plt.xlabel('Training Iteration', fontsize=16)
plt.ylabel('Reward', fontsize=16)
plt.tick_params(axis='both', labelsize=16)
plt.legend()
plt.tight_layout()
# plt.savefig(figurestorepath/'reward.svg', format='svg')
plt.savefig(figurestorepath/'reward.jpg', dpi=600, format='jpg')
plt.show()
```
