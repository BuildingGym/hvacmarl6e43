```python
from hvacmarl6e43.buildings import OfficeBuilding
from hvacmarl6e43.experiments import rl_experiments
from pathlib import Path
from hvacmarl6e43_notebooks import notebook_storage_base
import holoviews as hv
import matplotlib.pyplot as plt
hv.extension('matplotlib', 'bokeh')
hv.extension('bokeh')
import numpy as np

figurestorepath = Path(notebook_storage_base)
metrics = {
    name: experiment.get_results().metrics_train
    for name, experiment in rl_experiments.items()
}
```

```python
name_edit = {
    'monoagent': 'SAPPO-PBT (Special Case)',
    'multiagent': 'MAPPO-PBT (Proposed)',
}

for name in list(metrics.keys()):
    if name =='monoagent':
        metrics[name_edit[name]] = metrics[name]
        del metrics[name]
    elif name == 'multiagent':
        metrics[name_edit[name]] = metrics[name]
        del metrics[name]
```

```python
rewards = {
    'SAPPO-PBT (Special Case)':
        metrics['SAPPO-PBT (Special Case)']['env_runners/policy_reward_mean/default_policy']
        .dropna(),
    'MAPPO-PBT (Proposed)': 
        metrics['MAPPO-PBT (Proposed)'][[
            f'env_runners/hist_stats/policy_{zone_key}_reward' 
            for zone_key in OfficeBuilding.zone_keys
        ]]
        .dropna()
        .map(np.mean)
        .mean(axis='columns'),
}

colors = {
    'Baseline (23.0°C)': (239/255, 65/255, 67/255),  
    'Baseline (24.0°C)': (191/255, 30/255, 46/255),   
    'SAPPO-PBT (Special Case)': (115/255, 186/255, 214/255),     
    'MAPPO-PBT (Proposed)': (3/255, 50/255, 80/255),
}
```

```python
plt.figure(figsize=(10, 6))
%matplotlib inline
for label, reward_hist in rewards.items():
    x = reward_hist.index
    y = reward_hist.values
    plt.plot(x,y, label=label,color=colors[label])
    avg_reward = reward_hist.mean()
    plt.axhline(y=avg_reward, color=colors[label], linestyle='--', 
                label=f"{label} average ({avg_reward:.4f})")
plt.xlabel('Training Iteration', fontsize=16)
plt.ylabel('Reward', fontsize=16)
plt.tick_params(axis='both', labelsize=16)
plt.legend()
plt.tight_layout()
plt.savefig(figurestorepath/'reward.jpg', dpi=600, format='jpg')
plt.show()
```
