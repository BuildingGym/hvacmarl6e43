```python
import matplotlib.pyplot as plt
plt.rcParams.update({
    'backend': 'svg',
    'pgf.texsystem': 'lualatex',
    'pgf.rcfonts': False,
    'text.usetex': True,
    'font.family': 'serif',
    'font.serif': [],
})

# TODO plt.style.available
# plt.style.use('seaborn-v0_8-paper')

import matplotlib.axis as maxis
import matplotlib.dates as mdates

def mpl_autofmt_datetime(axis: maxis.Axis):
    axis.set_major_locator(locator := mdates.AutoDateLocator())
    axis.set_major_formatter(formatter := mdates.ConciseDateFormatter(axis.get_major_locator()))
    return locator, formatter

from matplotlib.patches import Patch

def mpl_shared_legend_patches(axes):
    return [
        Patch(color=color, label=label) 
        for (label, color) in dict.fromkeys([
            (label, handle.get_color())
            for ax in axes
            for handle, label in zip(*ax.get_legend_handles_labels())
        ])
    ]

from hvplot import hvPlot as Plot
import holoviews as hv
#hv.extension('bokeh', 'plotly', 'matplotlib')
hv.extension('matplotlib', 'bokeh')

import pandas as pd
```

```python
baseline_mapping = {
    '23.0': 'Baseline (23.0°C)',
    '24.0': 'Baseline (24.0°C)',
    'monoagent':'SAPPO-PBT (Special Case)',
    'multiagent':'MAPPO-PBT (Proposed)',
}

for old_name in list(datas.keys()):
    if old_name in baseline_mapping:
        new_name = baseline_mapping[old_name]
        datas[new_name] = datas[old_name]
        del datas[old_name]
```

```python
zone_map = {
    "EAST1": "ZONE2",
    "EAST":  "ZONE1",
    "WEST1": "ZONE4",
    "WEST":  "ZONE3",
}
for scheme_name, rooms_dict in datas.items():
    for room_id in list(rooms_dict.keys()):  
        for suffix, zone in zone_map.items():
            if suffix in room_id:
                new_room_id = room_id.replace(suffix, zone)
                rooms_dict[new_room_id] = rooms_dict.pop(room_id)
                break
```

```python
from collections import defaultdict

datas_eachroom = defaultdict(dict)
for name, data in datas.items():
    for room_id, df in data.items():
        df = pd.DataFrame(df)
        df['is_comfortable'] = df['pmv_abs'] <= .5
        datas_eachroom[room_id][name] = df
```

```python
from itertools import zip_longest

n_cols = 2
n_rows = round(len(datas_eachroom) / n_cols)
fig = plt.figure(figsize=(12, 8))
subfig_legend, subfig_data = fig.subfigures(
    nrows=2, ncols=1,
    height_ratios=(.1, .9),
)

axes = subfig_data.subplots(
    n_rows, 
    n_cols, 
    sharex=True, sharey=True, 
)

wide_locator = mdates.HourLocator(byhour=[0, 12]) 
safe_formatter = mdates.DateFormatter('%m-%d %H:%M') # 用空格代替 \n 防止报错

for ax, data in zip_longest(
    axes.flat, 
    datas_eachroom.items(),
    fillvalue=None,
):
    if data is None:
        ax.axis('off')
        continue
    
    room_id, room_datas = data
    if "EAST1" in room_id:
        room_title = room_id.replace("EAST1", "ZONE2")
    elif "EAST" in room_id:
        room_title = room_id.replace("EAST", "ZONE1")
    elif "WEST1" in room_id:
        room_title = room_id.replace("WEST1", "ZONE4")
    elif "WEST" in room_id:
        room_title = room_id.replace("WEST", "ZONE3")
    else:
        room_title = room_id
        
    for experiment_name, df in room_datas.items():
        sorted_df = df.sort_index().loc[(df.index >= start_date) & (df.index <= end_date)]
        ax.plot(sorted_df.index, sorted_df['pmv_abs'], label=str(experiment_name), alpha=0.75, color=colors[experiment_name])
    ax.set_title(str(room_title), loc='left')

for ax in axes.flat:
    ax.xaxis.set_major_locator(wide_locator)
    ax.xaxis.set_major_formatter(safe_formatter)
    ax.set_yticks(np.arange(0, 2, .5))
    ax.tick_params(axis='x', labelsize=10)
    ax.tick_params(axis='y', labelsize=14)
    
    ax.label_outer()

bottom_axes = axes[-1, :] if axes.ndim > 1 else axes
for ax in bottom_axes:
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha='right')

subfig_legend.legend(
    handles=(handles := mpl_shared_legend_patches(axes.flat)), 
    ncol=len(handles),
    loc='center',
    fontsize=14,
)

fig.supxlabel('Time', fontsize=14)
fig.supylabel('Absolute PMV', fontsize=14)
fig.tight_layout()

# TODO rm
fig.savefig(figurestorepath /'figure_time-v-pmv-abs.svg')
fig.savefig(figurestorepath /'figure_time-v-pmv-abs.jpg', dpi=600, bbox_inches='tight')
fig
```

```python
unified_index = pd.date_range(start=start_date, end=end_date, freq='10min')

comfort_datas_02 = dict()
comfort_datas_05 = dict()

for name, data in datas.items():
    df_02 = pd.DataFrame({
        room_id: df['pmv_abs'].le(0.2).reindex(unified_index, method='pad')  
        for room_id, df in data.items()
    })
    df_sum_02 = df_02.mean(axis='columns')
    comfort_datas_02[name] = df_sum_02
    
    df_05 = pd.DataFrame({
        room_id: df['pmv_abs'].le(0.5).reindex(unified_index, method='pad')
        for room_id, df in data.items()
    })
    df_sum_05 = df_05.mean(axis='columns')
    comfort_datas_05[name] = df_sum_05
```

```python
import matplotlib.ticker as mtick
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 6), sharex=True,constrained_layout=True)


for key, df in comfort_datas_02.items():
    ax1.plot(
        df.index, df, label=key,
        linestyle='-', alpha=0.75,
        color=colors[key],
    )
ax1.set_title('(a)$|\mathrm{PMV}|\leq0.2$',fontsize = 14)


for key, df in comfort_datas_05.items():
    ax2.plot(
        df.index, df, label=key,
        linestyle='-', alpha=0.75,
        color=colors[key],
    )
ax2.set_title('(b)$|\mathrm{PMV}|\leq0.5$',fontsize = 14)

for ax in [ax1, ax2]:
    mpl_autofmt_datetime(ax.xaxis)
    ax.xaxis.set_major_locator(mdates.HourLocator(byhour=[0, 6, 12, 18]))
    ax.xaxis.get_offset_text().set_visible(False)
    ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.))
    ax.tick_params(axis='x', labelsize=14)
    ax.tick_params(axis='y', labelsize=14)


handles, labels = ax1.get_legend_handles_labels()
fig.legend(handles, labels, loc='upper center', bbox_to_anchor=(0.5, 1.08), ncol=4, fontsize = 14)
fig.autofmt_xdate()
fig.supxlabel('Time',fontsize = 16)
fig.supylabel('Proportion of Rooms Meeting the Comfort Criteria',fontsize = 14)
fig.tight_layout()

fig.savefig(figurestorepath / 'figure_time-v-comfort-horizontal.jpg', 
            dpi=600, bbox_inches='tight', format='jpg')


plt.show()
```
