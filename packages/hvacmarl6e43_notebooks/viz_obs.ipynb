```python
import matplotlib.pyplot as plt
plt.rcParams.update({
    'backend': 'svg',
    'pgf.texsystem': 'lualatex',
    'pgf.rcfonts': False,
    'text.usetex': True,
    'font.family': 'serif',
    'font.serif': [],
})

# TODO plt.style.available
# plt.style.use('seaborn-v0_8-paper')

import matplotlib.axis as maxis
import matplotlib.dates as mdates

def mpl_autofmt_datetime(axis: maxis.Axis):
    axis.set_major_locator(locator := mdates.AutoDateLocator())
    axis.set_major_formatter(formatter := mdates.ConciseDateFormatter(axis.get_major_locator()))
    return locator, formatter

from matplotlib.patches import Patch

def mpl_shared_legend_patches(axes):
    return [
        Patch(color=color, label=label) 
        for (label, color) in dict.fromkeys([
            (label, handle.get_color())
            for ax in axes
            for handle, label in zip(*ax.get_legend_handles_labels())
        ])
    ]

from hvplot import hvPlot as Plot
import holoviews as hv
#hv.extension('bokeh', 'plotly', 'matplotlib')
hv.extension('matplotlib', 'bokeh')

import pandas as pd
```

```python
from pathlib import Path
from hvacmarl6e43_notebooks import notebook_storage_base

figurestorepath = Path(notebook_storage_base)
figurestorepath.mkdir(exist_ok=True)
```

```python
from hvacmarl6e43.experiments import baseline_experiments, rl_experiments

datas = {
    name: experiment.get_results().observations
    for name, experiment in {
        **baseline_experiments,
        **rl_experiments,
    }.items()
}

start_date = pd.to_datetime('2002-07-08').tz_localize('UTC')
end_date = pd.to_datetime('2002-07-13').tz_localize('UTC')
```

```python
hv.extension('bokeh')
datas_avg = dict()

for name, data in datas.items():
    df = pd.concat(data, axis='columns')
    df_avg = df.groupby(level=1, axis='columns').mean()
    datas_avg[name] = df_avg


datas_avg_elec_sorted = {
    key: df.sort_values(by='elec')
    for key, df in datas_avg.items()
}


hv.Overlay([
    Plot(df.loc[~((df['occupancy'] == 0) | (df['elec'] == 0))])
    .scatter(x='elec', y='pmv_abs', label=str(name))
    .opts(alpha=.8, size=1)
    for name, df in datas_avg_elec_sorted.items()
]).opts(width=800, height=500, legend_position='top_left')
```

```python
import matplotlib.pyplot as plt
%matplotlib inline
plt.figure(figsize=(7.5, 4))

for name, df in datas_avg_elec_sorted.items():
    df_filtered = df.loc[~((df['occupancy'] == 0) | (df['elec'] == 0))]
    
    plt.scatter(df_filtered['elec'], df_filtered['pmv_abs'], label=str(name), alpha=0.8, s=.5)


plt.xlabel('Elctricity(W)')
plt.ylabel('Absolute PMV')
legend = plt.legend()
plt.legend(markerscale=4.0)
plt.grid(True)
plt.savefig(figurestorepath /"elec_vs_PMV_1.svg",bbox_inches='tight')
```

```python
import numpy as np
from scipy.interpolate import interp1d


baseline_names = ['24.0', '23.5', '23.0', '22.5']
df_selected = df[df["name"].isin(baseline_names)].sort_values(by="elec_mean")
df_baseline_points = df[df["name"].isin(["monoagent", "multiagent"])].copy()
baseline_interp = interp1d(df_selected["elec_mean"], df_selected["pmv_abs_mean"], kind='linear', fill_value='extrapolate')

inverse_baseline_interp = interp1d(df_selected["pmv_abs_mean"], df_selected["elec_mean"], kind='linear', fill_value='extrapolate')
df_baseline_points["pmv_baseline"] = baseline_interp(df_baseline_points["elec_mean"])
df_baseline_points["elec_baseline"] = inverse_baseline_interp(df_baseline_points["pmv_abs_mean"])
df_baseline_points["vertical_distance"] = abs(df_baseline_points["pmv_abs_mean"] - df_baseline_points["pmv_baseline"])
df_baseline_points["horizontal_distance"] = abs(df_baseline_points["elec_mean"] - df_baseline_points["elec_baseline"])

plt.figure(figsize=(7.5, 4))
plt.scatter(df["elec_mean"], df["pmv_abs_mean"], s=15,color='blue', label='All Data')
plt.plot(df_selected["elec_mean"], df_selected["pmv_abs_mean"], linestyle='-', color='red', marker='o', markersize=4, label='Selected Points')

for _, row in df_baseline_points.iterrows():
    plt.plot([row["elec_mean"], row["elec_mean"]], [row["pmv_baseline"], row["pmv_abs_mean"]],
             color='gray', linestyle='dashed', linewidth=1.5)
    
    plt.plot([row["elec_mean"], row["elec_baseline"]], [row["pmv_abs_mean"], row["pmv_abs_mean"]],
             color='gray', linestyle='dashed', linewidth=1.5)
    
    plt.text(row["elec_mean"]-15_000, (row["pmv_abs_mean"] + row["pmv_baseline"]) / 2, 
             f"{row['vertical_distance']:.3f}", fontsize=7, ha='right', va='center', color='black')

    horizontal_distance_scaled = row["horizontal_distance"] / 1e5
    horizontal_label = r"${:.1f} \times 10^5$".format(horizontal_distance_scaled)
    plt.text((row["elec_mean"] + row["elec_baseline"]) / 2, row["pmv_abs_mean"]-0.005, 
             horizontal_label, fontsize=8, ha='center', va='top', color='black')

for _ , row in df.iterrows():
    if row["name"]=="monoagent" :
        plt.text(row["elec_mean"], row["pmv_abs_mean"], row["name"], fontsize=7, ha='right', color='blue')
        pass
    else:
        plt.text(row["elec_mean"], row["pmv_abs_mean"]- 0.01, row["name"], fontsize=7, ha='right', color='blue')

plt.xlabel("Wattage (Avg.)")
plt.ylabel("PMV (Absolute Avg.)")
# plt.legend()
# plt.savefig(figurestorepath /"figure:elec-mean-v-pmv-abs-mean.svg", dpi=600,bbox_inches='tight')
# plt.show()

```

```python
from collections import defaultdict

datas_eachroom = defaultdict(dict)
for name, data in datas.items():
    for room_id, df in data.items():
        df = pd.DataFrame(df)
        df['is_comfortable'] = df['pmv_abs'] <= .5
        datas_eachroom[room_id][name] = df
```

```python
from itertools import zip_longest

n_cols = 2
n_rows = round(len(datas_eachroom) / n_cols)
fig = plt.figure(figsize=(12, 6))
subfig_legend, subfig_data = fig.subfigures(
    nrows=2, ncols=1,
    height_ratios=(.15, .85),
)

axes = subfig_data.subplots(
    n_rows, 
    n_cols, 
    sharex=True, sharey=True, 
)
for ax, data in zip_longest(
    axes.flat, 
    datas_eachroom.items(),
    fillvalue=None,
):
    if data is None:
        ax.axis('off')
        continue
    
    room_id, room_datas = data
    if "EAST1" in room_id:
        room_title = room_id.replace("EAST1", "ZONE2")
    elif "EAST" in room_id:
        room_title = room_id.replace("EAST", "ZONE1")
    elif "WEST1" in room_id:
        room_title = room_id.replace("WEST1", "ZONE4")
    elif "WEST" in room_id:
        room_title = room_id.replace("WEST", "ZONE3")
    else:
        room_title = room_id
    for experiment_name, df in room_datas.items():
        sorted_df = df.sort_index().loc[(df.index >= start_date) & (df.index <= end_date)]
        ax.plot(sorted_df.index, sorted_df['pmv_abs'], label=str(experiment_name),alpha=0.75)
    ax.set_title(str(room_title), loc='left')
for ax in axes.flat:
    mpl_autofmt_datetime(ax.xaxis)
    ax.xaxis.set_major_locator(mdates.HourLocator(byhour=[0, 6, 12, 18]))
    ax.xaxis.get_offset_text().set_visible(False)
    ax.set_yticks(np.arange(0, 2, .5))
subfig_data.supxlabel('Time')
subfig_data.autofmt_xdate()
subfig_legend.legend(
    handles=(handles := mpl_shared_legend_patches(axes.flat)), 
    ncol=len(handles),
    loc='center',
)

fig.supxlabel('Time')
fig.supylabel('Absolute PMV')
fig.tight_layout()

fig.savefig(figurestorepath / 'figure:time-v-pmv_abs-per-zone.pgf')
# TODO rm
fig.savefig(figurestorepath /'figure:time-v-pmv-abs.svg')
fig
```
