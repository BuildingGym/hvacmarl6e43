```python
from hvacmarl6e43.experiments import baseline_experiments, rl_experiments

datas = {
    name: experiment.get_results().observations
    for name, experiment in {
        **baseline_experiments,
        **rl_experiments,
    }.items()
}


start_date = pd.to_datetime('2002-07-08').tz_localize('UTC')
end_date = pd.to_datetime('2002-07-13').tz_localize('UTC')

zoom_start_date = pd.to_datetime('2002-07-08').tz_localize('UTC')
zoom_end_date = pd.to_datetime('2002-07-11').tz_localize('UTC')

zoom_start_date_1 = pd.to_datetime('2002-07-07').tz_localize('UTC').replace(hour=7, minute=0, second=0)
zoom_end_date_1 = pd.to_datetime('2002-07-07').tz_localize('UTC').replace(hour=19, minute=0, second=0)
```

```python
baseline_mapping = {
    '23.0': 'Baseline (23.0°C)',
    '24.0': 'Baseline (24.0°C)',
    'monoagent':'SAPPO-PBT (Special Case)',
    'multiagent':'MAPPO-PBT (Proposed)',
}

for old_name in list(datas.keys()):
    if old_name in baseline_mapping:
        new_name = baseline_mapping[old_name]
        datas[new_name] = datas[old_name]
        del datas[old_name]
```

```python
zone_map = {
    "EAST1": "ZONE2",
    "EAST":  "ZONE1",
    "WEST1": "ZONE4",
    "WEST":  "ZONE3",
}
for scheme_name, rooms_dict in datas.items():
    for room_id in list(rooms_dict.keys()):  
        for suffix, zone in zone_map.items():
            if suffix in room_id:
                new_room_id = room_id.replace(suffix, zone)
                rooms_dict[new_room_id] = rooms_dict.pop(room_id)
                break
```

```python
hv.extension('bokeh')
datas_avg = dict()

for name, data in datas.items():
    df = pd.concat(data, axis='columns')
    df_avg = df.groupby(level=1, axis='columns').mean()
    datas_avg[name] = df_avg


datas_avg_elec_sorted = {
    key: df.sort_values(by='elec')
    for key, df in datas_avg.items()
}


hv.Overlay([
    Plot(df.loc[~((df['occupancy'] == 0) | (df['elec'] == 0))])
    .scatter(x='elec', y='pmv_abs', label=str(name))
    .opts(alpha=.8, size=1)
    for name, df in datas_avg_elec_sorted.items()
]).opts(width=800, height=500, legend_position='top_left')
```

```python
datas_avg_elec_sorted_rolling: dict[object, pd.DataFrame] = dict()
for key, df in datas_avg_elec_sorted.items():
    df = df.loc[~((df['occupancy'] == 0) | (df['elec'] == 0))]
    pmv_abs_rolling = df['pmv_abs'].rolling(window=10, min_periods=1)
    datas_avg_elec_sorted_rolling[key] = pd.DataFrame({
        'elec': df['elec'],
        'pmv_abs': df['pmv_abs'],
        'pmv_abs_rolling_mean': pmv_abs_rolling.mean(),
        'pmv_abs_rolling_min': pmv_abs_rolling.min(),
        'pmv_abs_rolling_max': pmv_abs_rolling.max(),
    })

fig = hv.Overlay([
    hv.Overlay([
        hv.Curve(
            (df['elec'], df['pmv_abs_rolling_mean']), 
            label=str(name),
        ).opts(alpha=.8),
        hv.Area(
            (df['elec'], df['pmv_abs_rolling_min'], df['pmv_abs_rolling_max']), 
            vdims=['y_min', 'y_max'], 
            label=str(name),
        )
        .opts(xlabel=r'Energy Consumption (W)', ylabel='Absolute PMV (Rolling Avg.)')
        .opts(alpha=.1, line_width=0, linewidth=0)
    ])
    for name, df in datas_avg_elec_sorted_rolling.items()
]).opts() #.opts(width=1_000, legend_position='right')
# hv.save(fig, figurestorepath / 'figure_elec-v-pmv_abs_rolling.svg')
fig
```

```python
comfort_datas_02_1 = dict()
comfort_datas_05_1 = dict()

# Iterate over the datas_avg dictionary
for name, data in datas_avg.items():
    # Filter the data where occupancy or electric usage is zero
    filtered_data = data.loc[~((data['occupancy'] == 0) | (data['elec'] == 0))]

    # Calculate the proportion of rooms where PMV <= 0.2
    proportion_02 = (filtered_data['pmv_abs'] <= 0.2).mean()
    comfort_datas_02_1[name] = proportion_02  # Store result for this data

    # Calculate the proportion of rooms where PMV <= 0.5
    proportion_05 = (filtered_data['pmv_abs'] <= 0.5).mean()
    comfort_datas_05_1[name] = proportion_05  # Store result for this data

# Calculate overall ratios for each method
overall_ratios = {}

for name, proportion_02 in comfort_datas_02_1.items():
    overall_ratios[name] = {
        'PMV <= 0.2': proportion_02,
        'PMV <= 0.5': comfort_datas_05_1[name]
    }

# Display the overall ratios
print("Overall ratio of rooms meeting the comfort criteria:")

for method, ratios in overall_ratios.items():
    print(f"\n{method}:")
    print(f"  PMV <= 0.2: {ratios['PMV <= 0.2']:.2%}")
    print(f"  PMV <= 0.5: {ratios['PMV <= 0.5']:.2%}")
```
