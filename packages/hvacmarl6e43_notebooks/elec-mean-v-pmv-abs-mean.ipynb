```python
import matplotlib.pyplot as plt
plt.rcParams.update({
    'backend': 'svg',
    'pgf.texsystem': 'lualatex',
    'pgf.rcfonts': False,
    'text.usetex': True,
    'font.family': 'serif',
    'font.serif': [],
})

# TODO plt.style.available
# plt.style.use('seaborn-v0_8-paper')

import matplotlib.axis as maxis
import matplotlib.dates as mdates

def mpl_autofmt_datetime(axis: maxis.Axis):
    axis.set_major_locator(locator := mdates.AutoDateLocator())
    axis.set_major_formatter(formatter := mdates.ConciseDateFormatter(axis.get_major_locator()))
    return locator, formatter

from matplotlib.patches import Patch

def mpl_shared_legend_patches(axes):
    return [
        Patch(color=color, label=label) 
        for (label, color) in dict.fromkeys([
            (label, handle.get_color())
            for ax in axes
            for handle, label in zip(*ax.get_legend_handles_labels())
        ])
    ]

from hvplot import hvPlot as Plot
import holoviews as hv
#hv.extension('bokeh', 'plotly', 'matplotlib')
hv.extension('matplotlib', 'bokeh')

import pandas as pd
```

```python
from pathlib import Path
from hvacmarl6e43_notebooks import notebook_storage_base

figurestorepath = Path(notebook_storage_base)
figurestorepath.mkdir(exist_ok=True)
```

```python
from hvacmarl6e43.experiments import baseline_experiments, rl_experiments

datas = {
    name: experiment.get_results().observations
    for name, experiment in {
        **baseline_experiments,
        **rl_experiments,
    }.items()
}


start_date = pd.to_datetime('2002-07-08').tz_localize('UTC')
end_date = pd.to_datetime('2002-07-13').tz_localize('UTC')
```

```python
hv.extension('bokeh')
datas_avg = dict()

for name, data in datas.items():
    df = pd.concat(data, axis='columns')
    df_avg = df.groupby(level=1, axis='columns').mean()
    datas_avg[name] = df_avg
```

```python
df_metrics_avg = pd.DataFrame([
    {
        'name': str(name),
        'elec_mean': df[df['occupancy'] != 0]['elec'].mean(),
        'pmv_abs_mean': df[df['occupancy'] != 0]['pmv_abs'].mean(),
    }
    for name, df in datas_avg.items()
])
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d
from matplotlib.patches import FancyArrowPatch
%matplotlib inline

colors = {
    'Baseline (23.0°C)': (239/255, 65/255, 67/255),
    'Baseline (24.0°C)': (191/255, 30/255, 46/255),
    'SAPPO-PBT (Spcial Case)': (115/255, 186/255, 214/255),
    'MAPPO-PBT (Proposed)': (3/255, 50/255, 80/255),
}

# 原有的数据处理
df_selected = df_metrics_avg[~df_metrics_avg["name"].isin(["monoagent", "multiagent"])].sort_values(by="elec_mean")

df_baseline_points = df_metrics_avg[df_metrics_avg["name"].isin(["monoagent", "multiagent"])].copy()

# 添加新的baseline点
baseline_23_row = df_metrics_avg[df_metrics_avg["name"] == "23.000000000000007"].copy()
baseline_24_row = df_metrics_avg[df_metrics_avg["name"] == "24.00000000000002"].copy()

baseline_23_row["name"] = "Baseline (23.0°C)"
baseline_24_row["name"] = "Baseline (24.0°C)"

# 合并所有需要标注的点
all_special_points = pd.concat([df_baseline_points, baseline_23_row, baseline_24_row], ignore_index=True)

baseline_interp = interp1d(df_selected["elec_mean"], df_selected["pmv_abs_mean"], kind='linear', fill_value='extrapolate')
inverse_baseline_interp = interp1d(df_selected["pmv_abs_mean"], df_selected["elec_mean"], kind='linear', fill_value='extrapolate')

all_special_points["pmv_baseline"] = baseline_interp(all_special_points["elec_mean"])
all_special_points["elec_baseline"] = inverse_baseline_interp(all_special_points["pmv_abs_mean"])
all_special_points["vertical_distance"] = abs(all_special_points["pmv_abs_mean"] - all_special_points["pmv_baseline"])
all_special_points["horizontal_distance"] = abs(all_special_points["elec_mean"] - all_special_points["elec_baseline"])

# 让baseline点浮于线上
for idx, row in all_special_points.iterrows():
    if row["name"] in ["Baseline (23.0°C)", "Baseline (24.0°C)"]:
        all_special_points.loc[idx, "pmv_abs_mean"] = row["pmv_baseline"]

plt.figure(figsize=(10, 6))

# Create a smoother baseline curve
x_smooth = np.linspace(df_selected["elec_mean"].min(), df_selected["elec_mean"].max(), 100)
y_smooth = baseline_interp(x_smooth)

# Plot baseline as a smooth curve without markers
plt.plot(
    x_smooth, y_smooth,
    linestyle='-', color='red', linewidth=2, label='Baseline', zorder=1
)

for _, row in all_special_points.iterrows():
    # 确定点的名称和颜色
    if row["name"] == "monoagent":
        agent_name = "SAPPO-PBT (Spcial Case)"
    elif row["name"] == "multiagent":
        agent_name = "MAPPO-PBT (Proposed)"
    else:
        agent_name = row["name"]  # Baseline (23.0°C) 或 Baseline (24.0°C)
    
    color = colors[agent_name]
    
    # 只为monoagent和multiagent创建箭头
    if row["name"] in ["monoagent", "multiagent"]:
        # Create vertical arrow for PMV improvement
        vertical_arrow = FancyArrowPatch(
            (row["elec_mean"], row["pmv_baseline"]),
            (row["elec_mean"], row["pmv_abs_mean"]),
            arrowstyle='->', color='gray', linewidth=1.5,
            mutation_scale=15, shrinkA=0, shrinkB=0
        )
        plt.gca().add_patch(vertical_arrow)
        
        # Create horizontal arrow for energy consumption improvement
        horizontal_arrow = FancyArrowPatch(
            (row["elec_mean"], row["pmv_abs_mean"]),
            (row["elec_baseline"], row["pmv_abs_mean"]),
            arrowstyle='<-', color='gray', linewidth=1.5,
            mutation_scale=15, shrinkA=0, shrinkB=0
        )
        plt.gca().add_patch(horizontal_arrow)
        
        # Add vertical distance label
        plt.text(
            row["elec_mean"] - 15000, 
            (row["pmv_abs_mean"] + row["pmv_baseline"]) / 2,
            f"{row['vertical_distance']:.3f}", fontsize=12, ha='right', va='center', color='black',
            bbox=dict(facecolor='white', alpha=0.7, edgecolor='none', pad=1)
        )
        
        # Add horizontal distance label
        horizontal_distance_scaled = row["horizontal_distance"] / 1e5
        horizontal_label = r"${:.1f} \times 10^5$".format(horizontal_distance_scaled)
        plt.text(
            (row["elec_mean"] + row["elec_baseline"]) / 2, 
            row["pmv_abs_mean"] - 0.01,
            horizontal_label, fontsize=12, ha='center', va='top', color='black',
            bbox=dict(facecolor='white', edgecolor='none', pad=1)
        )
    
    # Plot the algorithm point
    plt.scatter(row["elec_mean"], row["pmv_abs_mean"], color=color, s=50, label=agent_name, zorder=2)
    
    # 确定标签位置和坐标信息
    elec_mean_scaled = row['elec_mean']/ 1e7
    elec_mean_label = r"${:.3f} \times 10^7$".format(elec_mean_scaled)
    coord_text = f"({elec_mean_label}, {row['pmv_abs_mean']:.4f})"
    
    # 根据点的类型调整标签位置
    if row["name"] in ["Baseline (23.0°C)", "Baseline (24.0°C)"]:
        # Baseline点的标签放在右边
        plt.text(
            row["elec_mean"] + 40000, 
            row["pmv_abs_mean"] , 
            agent_name, 
            fontsize=14, 
            ha='left', 
            va='bottom',
            color=color,
            bbox=dict(facecolor='white', edgecolor='none', pad=1)
        )
        # 坐标信息放在名字下面
        plt.text(
            row["elec_mean"] + 150000, 
            row["pmv_abs_mean"] - 0.005, 
            coord_text, 
            fontsize=14, 
            ha='left', 
            va='top',
            color='black',
            bbox=dict(facecolor='white', edgecolor='none', pad=1)
        )
    else:
        # SAPPO-PBT和MAPPO-PBT的标签放在左边（原位置）
        plt.text(
            row["elec_mean"] - 30000, 
            row["pmv_abs_mean"] - 0.01, 
            agent_name, 
            fontsize=14, 
            ha='right', 
            va='top',
            color=color,
            bbox=dict(facecolor='white', edgecolor='none', pad=1)
        )
        # 坐标信息放在名字下面
        plt.text(
            row["elec_mean"] - 30000, 
            row["pmv_abs_mean"] - 0.028, 
            coord_text, 
            fontsize=14, 
            ha='right', 
            va='top',
            color='black',
            bbox=dict(facecolor='white', edgecolor='none', pad=1)
        )

plt.xlabel("Energy Consumption (W)", fontsize=14)
plt.ylabel("Absolute PMV (Avg.)", fontsize=14)
plt.tick_params(axis='x', labelsize=14)
plt.tick_params(axis='y', labelsize=14)

handles, labels = plt.gca().get_legend_handles_labels()
from collections import OrderedDict
by_label = OrderedDict(zip(labels, handles))
plt.legend(by_label.values(), by_label.keys(),fontsize = 14)

plt.grid(True)
plt.savefig(figurestorepath /"figure:elec-mean-v-pmv-abs-mean.jpg", dpi=600, bbox_inches='tight')
plt.tight_layout()
plt.show()
```
